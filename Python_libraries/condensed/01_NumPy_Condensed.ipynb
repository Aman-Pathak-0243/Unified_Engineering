## 1. What NumPy is
NumPy's `ndarray` stores same-typed values in an n-dimensional structure and supports fast vectorized operations.

```python
import numpy as np
```

````python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print(a + b)       # element-wise
print(a * 2)       # scalar broadcasting
print(a.dtype, a.shape, a.ndim, a.size)
````

## 2. Create arrays
Know these constructors: `array`, `arange`, `linspace`, `zeros`, `ones`, `full`, `eye`.

````python
print(np.arange(0, 10, 2))       # stop excluded
print(np.linspace(0, 1, 5))      # fixed number of points, endpoint included
print(np.zeros((2, 3)))
print(np.ones(4, dtype=int))
print(np.full((2, 2), 7))
print(np.eye(3))
````

## 3. Indexing, slicing, reshaping
For 2D arrays: `arr[row, column]`; use `:` to select a full axis. Slices usually create **views**; use `.copy()` when you need independence.

````python
x = np.arange(1, 13).reshape(3, 4)
print(x)
print(x[1, 2])       # row 1, col 2
print(x[:, -1])      # last column
print(x[:2, 1:3])    # submatrix
print(x.reshape(2, 6))
print(x.T)           # transpose
view = x[:2, :2]
copy = x[:2, :2].copy()
````

## 4. Vectorized math and aggregations
Common methods: `sum`, `mean`, `std`, `min`, `max`, `argmin`, `argmax`, `cumsum`. **Axis rule:** `axis=0` reduces rows (column-wise result); `axis=1` reduces columns (row-wise result).

````python
marks = np.array([[70, 80, 90],
                   [60, 75, 85],
                   [95, 88, 92]])
print(marks.mean(axis=0))  # subject averages
print(marks.mean(axis=1))  # student averages
print(marks.max(), marks.argmax())
print(np.sqrt(marks))
print(marks.sum(axis=1, keepdims=True))
````

## 5. Broadcasting
Shapes are compared from the right. Dimensions must match or one must be `1`. This lets a 1D row of values operate across a 2D array.

````python
scores = np.array([[70, 80, 90], [60, 75, 85]])
max_marks = np.array([100, 100, 100])
percent = scores / max_marks * 100
centered = scores - scores.mean(axis=0)
print(percent)
print(centered)
````

## 6. Boolean masks, filtering, and conditional selection
Use `&`, `|`, `~` with parenthesized conditions—not Python's `and`/`or`.

````python
arr = np.array([12, 5, 18, 7, 25, 10])
print(arr[arr > 10])
print(arr[(arr >= 10) & (arr < 20)])
print((arr > 10).sum())          # count matches
print(np.where(arr >= 10, arr, 0))
````

## 7. Sorting and unique values
`sort` returns sorted values; `argsort` returns their indices. `unique` can return counts.

````python
v = np.array([30, 10, 20, 10, 40])
print(np.sort(v))
print(np.argsort(v))
print(np.unique(v, return_counts=True))
````

## 8. Linear algebra essentials
`*` is element-wise multiplication; `@` is matrix multiplication. Prefer `np.linalg.solve(A, b)` over explicitly calculating an inverse.

````python
A = np.array([[2., 1.], [1., 3.]])
b = np.array([110., 180.])
print(A @ A)
print(np.linalg.solve(A, b))
print(np.linalg.det(A))
print(np.linalg.norm([3., 4.]))
````

## 9. Reproducible random numbers
Use `default_rng(seed)` for reproducible results.

````python
rng = np.random.default_rng(42)
print(rng.integers(0, 10, size=5))
print(rng.normal(loc=0, scale=1, size=5))
print(rng.choice([1, 2, 3], size=5))
````

## 10. Common pitfalls
- `and`/`or` on arrays → use `&`/`|` and parentheses.
- Shape mismatch → inspect `.shape`; reshape/add axes intentionally.
- Integer arrays can truncate assigned decimals → choose a float dtype.
- Boolean/fancy indexing generally returns a copy; basic slicing generally returns a view.
- Prefer vectorized expressions over Python loops for bulk numerical operations.

````python
# Pitfall examples
a = np.array([1, 2, 3], dtype=float)
a[0] = 1.5
print(a)

budget = np.array([[100, 200, 300], [400, 500, 600]])
adjustment = np.array([10, 20])  # one value per row
print(budget + adjustment[:, None])
````
