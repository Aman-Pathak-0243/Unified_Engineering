# Matplotlib — Condensed Practical Notes

A focused reference for ML/data analysis: chart selection, plotting, customization, layouts, statistical views, and saving figures.

## 1. Setup + Figure/Axes
Use the object-oriented API (`fig, ax = plt.subplots()`) for reusable, multi-plot code. `plt.*` is fine for quick exploration.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

x = np.arange(1, 6)
y = np.array([2, 4, 3, 7, 6])
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(x, y, marker='o')
ax.set(title='Example trend', xlabel='X', ylabel='Y')
plt.show()
```

## 2. Core chart types
- **Line:** ordered/continuous trends.
- **Scatter:** relationship/correlation between two numeric variables.
- **Bar:** compare categories.
- **Histogram:** distribution of one numeric variable.
- **Box plot:** compare distributions/spread and outliers.

```python
rng = np.random.default_rng(42)

# Line
fig, ax = plt.subplots(); ax.plot(np.arange(10), rng.normal(size=10), marker='o'); ax.set_title('Trend'); plt.show()

# Scatter
fig, ax = plt.subplots(); ax.scatter(rng.normal(size=100), rng.normal(size=100), alpha=.6); ax.set_title('Relationship'); plt.show()

# Bar
fig, ax = plt.subplots(); ax.bar(['A', 'B', 'C'], [12, 18, 9]); ax.set_title('Category comparison'); plt.show()

# Histogram
fig, ax = plt.subplots(); ax.hist(rng.normal(size=1000), bins=25, edgecolor='black'); ax.set_title('Distribution'); plt.show()
```

## 3. Labels, legend, grid, limits
Always label axes with units where applicable. Add a legend when multiple series appear.

```python
x = np.arange(1, 7)
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(x, x**2, marker='o', label='x²')
ax.plot(x, x**1.5, marker='s', label='x^1.5')
ax.set(title='Comparing series', xlabel='Input', ylabel='Output')
ax.set_ylim(0, 40)
ax.grid(True, alpha=.3)
ax.legend()
plt.show()
```

## 4. Multiple plots (subplots)
`plt.subplots(rows, columns, ...)` returns a Figure and an array of Axes. Use `ax[i]` / `axs[row, col]` to address each plot.

```python
fig, axs = plt.subplots(1, 2, figsize=(9, 3.5))
axs[0].plot(x, x**2); axs[0].set_title('Line')
axs[1].bar(['A', 'B', 'C'], [3, 7, 5]); axs[1].set_title('Bars')
fig.tight_layout()
plt.show()
```

## 5. Statistical plots
Use box plots to compare groups; error bars to show uncertainty. State what the error bars represent (SD, SE, or confidence interval).

```python
groups = [rng.normal(0, 1, 100), rng.normal(1, 1.5, 100), rng.normal(2, .8, 100)]
fig, ax = plt.subplots(); ax.boxplot(groups, tick_labels=['G1', 'G2', 'G3']); ax.set_ylabel('Value'); plt.show()

means = [10, 13, 12]; errors = [1.2, 1.8, 1.0]
fig, ax = plt.subplots(); ax.errorbar(['A', 'B', 'C'], means, yerr=errors, fmt='o', capsize=4); ax.set_ylabel('Mean ± error'); plt.show()
```

## 6. Plotting Pandas data
Pandas `.plot()` is convenient for quick EDA; Matplotlib Axes methods provide finer control.

```python
df = pd.DataFrame({'month': ['Jan', 'Feb', 'Mar', 'Apr'], 'sales': [12, 16, 14, 21], 'cost': [8, 9, 10, 13]})
ax = df.plot(x='month', y=['sales', 'cost'], marker='o', figsize=(6, 3.5), title='Sales vs cost')
ax.set_ylabel('Amount')
plt.show()
```

## 7. Save figures
Save before/after `show()` depending on backend; use `bbox_inches='tight'` to avoid clipped labels. PNG is common for reports; SVG/PDF are vector formats.

```python
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(x, x**2, marker='o')
ax.set(title='Saved chart', xlabel='X', ylabel='X squared')
fig.savefig('chart.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close(fig)  # useful in loops / notebooks
```

## 8. Avoid misleading charts
- Bar charts should generally start at zero.
- Use consistent scales when comparing panels.
- Avoid unnecessary 3D effects and decoration.
- Make color encode a meaningful distinction; use labels/markers as additional cues.
- Show sample size and uncertainty when they matter.

## 9. Common mistakes
- **Figure vs Axes:** `fig` is the whole canvas; `ax` is an individual plot.
- **Mismatched lengths:** x and y must have compatible lengths.
- **Missing legend:** provide `label=` and call `ax.legend()`.
- **Plots accumulating:** create/close figures in loops (`plt.subplots()`, `plt.close(fig)`).
- **Clipped labels:** use `fig.tight_layout()` or `bbox_inches='tight'` when saving.

## Quick reference
| Need | Use |
|---|---|
| Trend | `ax.plot(x, y)` |
| Relationship | `ax.scatter(x, y)` |
| Category comparison | `ax.bar(labels, values)` |
| Distribution | `ax.hist(data, bins=...)` |
| Group distributions | `ax.boxplot(groups)` |
| Uncertainty | `ax.errorbar(x, y, yerr=...)` |
| Layout | `plt.subplots(r, c)` |
| Labels/title | `ax.set(title=..., xlabel=..., ylabel=...)` |
| Save | `fig.savefig(path, dpi=300, bbox_inches='tight')` |
